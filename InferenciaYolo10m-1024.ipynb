```python
# ===================================================================
# 🦁 INFERENCIA TILED 1024x1024 CON YOLOv10 — PRED vs REAL y EXACTITUD
# ===================================================================

import cv2
import os
import json
import torch
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import gridspec
from math import ceil

from ultralytics import YOLO

# ------------------------------------------------------------
# 0. CONFIGURACIÓN GLOBAL
# ------------------------------------------------------------
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("🔥 Using:", DEVICE)

YOLO_WEIGHTS = "/content/drive/MyDrive/maia_yolov10_Exp1/tiles_1024_jpg_q100/weights/best.pt"
GT_JSON_BIG  = "/content/drive/MyDrive/general_dataset/groundtruth/json/big_size/test_big_size_A_B_E_K_WH_WB.json"

TILE_SIZE = 1024
CONF_THR = 0.25
IOU_THR = 0.5

print("🔄 Cargando YOLOv10...")
model = YOLO(YOLO_WEIGHTS)
model.to(DEVICE)
CLASSES = model.names
print("✔️ Clases:", CLASSES)


# ============================================================
# 1. UTILIDADES
# ============================================================
def generate_fixed_tiles_1024(img):
    """Genera tiles de 1024x1024 para toda la imagen."""
    h, w = img.shape[:2]

    num_x = ceil(w / TILE_SIZE)
    num_y = ceil(h / TILE_SIZE)

    tiles = []
    coords = []

    for iy in range(num_y):
        for ix in range(num_x):
            x1 = ix * TILE_SIZE
            y1 = iy * TILE_SIZE
            x2 = min(x1 + TILE_SIZE, w)
            y2 = min(y1 + TILE_SIZE, h)

            tile = img[y1:y2, x1:x2]
            tiles.append(tile)
            coords.append((ix, iy, x1, y1, x2, y2))

    return tiles, coords, num_x, num_y


def draw_grid_1024(img, num_x, num_y):
    """Dibuja líneas 1024x1024 sobre la imagen."""
    g = img.copy()
    h, w = g.shape[:2]

    for ix in range(1, num_x):
        x = ix * TILE_SIZE
        cv2.line(g, (x, 0), (x, h), (0, 255, 255), 2)

    for iy in range(1, num_y):
        y = iy * TILE_SIZE
        cv2.line(g, (0, y), (w, y), (0, 255, 255), 2)

    return g


@torch.no_grad()
def infer_tile(frame):
    """Inferencia YOLO en tile."""
    r = model.predict(
        frame,
        conf=CONF_THR,
        iou=IOU_THR,
        imgsz=TILE_SIZE,
        device=DEVICE,
        verbose=False
    )[0]

    if not r.boxes:
        return np.zeros((0,4)), np.zeros((0,))

    boxes = r.boxes.xyxy.cpu().numpy()
    labels = r.boxes.cls.cpu().numpy().astype(int)

    return boxes, labels


def get_gt_counts_for_image(gt_file, filename):
    with open(gt_file, "r", encoding="utf-8") as f:
        data = json.load(f)

    img_id = None
    for im in data["images"]:
        if im["file_name"] == filename:
            img_id = im["id"]
            break

    if img_id is None:
        print("⚠️ No hay GT para:", filename)
        return {}

    cat_map = {c["id"]: c["name"] for c in data["categories"]}

    counts = {}
    for ann in data["annotations"]:
        if ann["image_id"] == img_id:
            cname = cat_map[ann["category_id"]]
            counts[cname] = counts.get(cname, 0) + 1

    return counts


def draw_boxes(img, boxes, labels, names):
    COLORS = {
        "Alcelaphinae": (0,0,180),
        "Buffalo": (255,200,150),
        "Kob": (0,140,255),
        "Warthog": (0,165,255),
        "Waterbuck": (128,0,255),
        "Elephant": (50,220,50)
    }

    out = img.copy()
    for b, lab in zip(boxes, labels):
        x1,y1,x2,y2 = map(int,b)
        cname = names[int(lab)]
        col = COLORS.get(cname, (0,255,0))

        cv2.rectangle(out, (x1,y1), (x2,y2), col, 2)
        cv2.putText(out, cname, (x1, max(y1-5,15)), cv2.FONT_HERSHEY_SIMPLEX, 0.6, col, 2)
    return out


# ============================================================
# 2. INFERENCIA COMPLETA
# ============================================================
def infer_big_image(path_img):
    big = cv2.imread(path_img)
    if big is None:
        raise FileNotFoundError(path_img)

    fname = os.path.basename(path_img)

    tiles, coords, num_x, num_y = generate_fixed_tiles_1024(big)
    print(f"🧩 Tiles generados: {len(tiles)} ({num_x} x {num_y})")

    grid_img = draw_grid_1024(big, num_x, num_y)

    # Mostrar cuadrícula
    plt.figure(figsize=(12,8))
    plt.imshow(cv2.cvtColor(grid_img, cv2.COLOR_BGR2RGB))
    plt.title("Cuadrícula 1024x1024")
    plt.axis("off")
    plt.show()

    print("🚀 Iniciando inferencia tile por tile...")

    total_counts = {}
    per_tile_counts = {}
    tiles_boxed = []

    for i, (tile, (ix, iy, x1, y1, x2, y2)) in enumerate(zip(tiles, coords), start=1):

        boxes, labels = infer_tile(tile)

        # guardar conteos
        local = {}
        for lab in labels:
            cname = CLASSES[lab]
            total_counts[cname] = total_counts.get(cname, 0) + 1
            local[cname] = local.get(cname, 0) + 1

        tile_name = f"Tile {iy}-{ix}"
        per_tile_counts[tile_name] = local

        tiles_boxed.append((tile_name, draw_boxes(tile, boxes, labels, CLASSES)))

    # Mostrar tiles
    n = len(tiles_boxed)
    cols_vis = min(num_x, n)
    rows_vis = ceil(n / cols_vis)

    plt.figure(figsize=(4 * cols_vis, 4 * rows_vis))
    gs = gridspec.GridSpec(rows_vis, cols_vis)

    for i, (name, timg) in enumerate(tiles_boxed):
        ax = plt.subplot(gs[i // cols_vis, i % cols_vis])
        ax.imshow(cv2.cvtColor(timg, cv2.COLOR_BGR2RGB))
        ax.set_title(name)
        ax.axis("off")

    plt.tight_layout()
    plt.show()

    # Mostrar conteo por tile
    print("\n📌 Conteo por tile:")
    for k,v in per_tile_counts.items():
        print(f" {k}: {v}")

    # GT
    gt_counts = get_gt_counts_for_image(GT_JSON_BIG, fname)

    # GT vs Pred
    if gt_counts:
        all_species = sorted(set(gt_counts) | set(total_counts))
        gt_vals = [gt_counts.get(s,0) for s in all_species]
        pr_vals = [total_counts.get(s,0) for s in all_species]

        y = np.arange(len(all_species))
        width = 0.35

        plt.figure(figsize=(10,6))
        plt.barh(y-width/2, gt_vals, width, label="GT")
        plt.barh(y+width/2, pr_vals, width, label="Pred")

        plt.yticks(y, all_species)
        plt.legend()
        plt.title("GT vs YOLOv10 — Conteo Global")
        plt.show()

        # exactitud
        sum_gt = sum(gt_vals)
        if sum_gt > 0:
            abs_err = sum(abs(p-g) for p,g in zip(pr_vals, gt_vals))
            acc = 1 - abs_err / sum_gt
            print(f"\n🎯 Exactitud global: {acc*100:.2f}%")

    return total_counts, per_tile_counts, gt_counts


# ============================================================
# 3. EJECUCIÓN /content/drive/MyDrive/general_dataset/test/L_09_05_16_DSC10034.JPG
#               /content/drive/MyDrive/general_dataset/train/S_10_05_16_DSC00794.JPG
#              /content/drive/MyDrive/general_dataset/test/01802f75da35434ab373569fffc1fd65a3417aef.JPG
# ============================================================
path = input("Ruta de la imagen BIG: ").strip()
pred, per_tile, gt = infer_big_image(path)

print("\n✔️ Predicción total:", pred)
print("✔️ Conteo GT:", gt)
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

COLORS = {
    "Alcelaphinae": (0,0,180),
    "Buffalo": (255,200,150),
    "Kob": (0,140,255),
    "Warthog": (0,165,255),
    "Waterbuck": (128,0,255),
    "Elephant": (50,220,50)
}

json_path = "/content/drive/MyDrive/general_dataset/groundtruth/json/big_size/train_big_size_A_B_E_K_WH_WB.json"
train_folder = "/content/drive/MyDrive/general_dataset/train"


MAX_IMAGES = 15    # límite solicitado
GRID_ROWS = 5
GRID_COLS = 3


# ============================
# 1. Cargar JSON
# ============================
with open(json_path, "r") as f:
    data = json.load(f)

images = {img["id"]: img for img in data["images"]}
annotations = data["annotations"]
categories = {cat["id"]: cat["name"] for cat in data["categories"]}

pred_species = set(pred.keys())


# ============================
# 2. Filtrar anotaciones de especies buscadas
# ============================
image_matches = {}

for ann in annotations:
    cat_name = categories[ann["category_id"]]
    if cat_name in pred_species:
        img_id = ann["image_id"]
        if img_id not in image_matches:
            image_matches[img_id] = {
                "count_total": 0,
                "per_species": {sp: 0 for sp in pred_species},
                "annotations": []
            }
        image_matches[img_id]["count_total"] += 1
        image_matches[img_id]["per_species"][cat_name] += 1
        image_matches[img_id]["annotations"].append(ann)


# ============================
# 3. Ordenar imágenes por mayor coincidencia
# ============================
ordered = sorted(
    image_matches.items(),
    key=lambda x: x[1]["count_total"],
    reverse=True
)

# Tomar solo las primeras 15
ordered = ordered[:MAX_IMAGES]


# ============================
# 4. Crear mosaico 3×5
# ============================
fig, axs = plt.subplots(GRID_ROWS, GRID_COLS, figsize=(90, 60))
axs = axs.flatten()

for ax in axs:
    ax.axis("off")

for idx, (img_id, info) in enumerate(ordered):
    ax = axs[idx]

    img_info = images[img_id]
    file_name = img_info["file_name"]
    img_path = os.path.join(train_folder, file_name)

    if not os.path.exists(img_path):
        print(f"❌ Imagen no encontrada: {img_path}")
        continue

    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    ax.imshow(img, aspect='auto')

    ax.set_title(
        f"{file_name}\nCoincidencias: {info['count_total']} | {info['per_species']}",
        fontsize=16
    )

    # Dibujar bounding boxes con colores
    for ann in info["annotations"]:
        x, y, w, h = ann["bbox"]
        cat_name = categories[ann["category_id"]]
        color = COLORS.get(cat_name, (255, 255, 255))

        # Caja
        rect = patches.Rectangle(
            (x, y), w, h,
            linewidth=2,
            edgecolor=[c/255 for c in color],
            facecolor='none'
        )
        ax.add_patch(rect)

        # Etiqueta
        ax.text(
            x, y - 3,
            cat_name,
            color='white',
            fontsize=14,
            backgroundcolor=[c/255 for c in color]
        )

plt.tight_layout(pad=0.1)
plt.subplots_adjust(
    left=0.02,
    right=0.98,
    top=0.98,
    bottom=0.02,
    wspace=0.02,
    hspace=0.02
)

plt.tight_layout()
plt.show()
```
