```python
from ultralytics import YOLO

# 1. Carga modelo base (ajusta a n/s/m/l según tu GPU)
model = YOLO("yolov10m.pt")

# 2. Paths en Drive
PROJECT_DIR = "/content/drive/MyDrive/maia_yolo_train_960"
RUN_NAME    = "yolov10m_960_tiles_v1"

model.train(
    data="/content/drive/MyDrive/general_dataset/yolo_tiles_960_menos/data.yaml",
    imgsz=960,
    epochs=100,
    batch=8,                     # ajusta según VRAM
    optimizer="sgd",
    lr0=0.01,
    warmup_epochs=3,

    # más estable para fauna pequeña
    mosaic=0.0,
    mixup=0.0,

    # early stopping
    patience=50,

    # ✅ todo se guarda en Drive
    project=PROJECT_DIR,
    name=RUN_NAME,

    # ✅ checkpoints
    save=True,        # guarda best.pt y last.pt
    save_period=10,   # además guarda cada 10 epochs

    pretrained=True,
)

```
